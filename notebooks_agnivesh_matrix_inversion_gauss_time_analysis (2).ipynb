```python
## code for matrix inversion using gauss jordan elimination method
import time
import copy
def invSolve(A1):
    A=copy.deepcopy(A1)
    ## same assumptions as before, since we are not allowed to use an in-built packages, we just assume a matrix is a list of lists, since we must perform inversion, this solution only holds for nxn matrices
    ## we inverse using the adjoint matrix method or gauss jordan elimination method
    n=len(A)
    I = [[0]*n for _ in range(n)]
    for i in range(n):
        I[i][i]=1
    ## We thus have our identity matrix to adjoint matrix ready
    for i in range(n):
        ## we know any row/column in excess of the rank automatically becomes zero in our process of GE as they are not independent and hence expressible as a linear combination of the independent row so they take care of themselves by becoming zero and being swapped out to the very end of the rows
        ## we have so far ensure that the first element of first row in the original system is no longer a zero, we will now proceed with the process of elimination and keep our strategy of partial pivoting everytime
        # normalize row i
        # pivoting
        if A[i][i] == 0:
            t = i + 1
            while t<n and A[t][i]==0:
                t+=1
            if t==n:
                raise ValueError("Matrix is singular")
            A[i],A[t]= A[t],A[i]
            I[i],I[t]=I[t],I[i]
        # normalize pivot row
        pivot=A[i][i]
        for k in range(n):
            A[i][k]/=pivot
            I[i][k]/=pivot
        for j in range(i+1,n): ## row loop
            multiple=A[j][i]/A[i][i]
            for k in range(0,n): ## column loop
                A[j][k]=A[j][k]-multiple*A[i][k]
                I[j][k]=I[j][k]-multiple*I[i][k]
        ## we are done with forward elimination, now we perform backward elimination
    for i in range(n-1, -1, -1):
         for j in range(i-1, -1, -1):
             ## we just eliminate the trailing zeroes
          multiple = A[j][i]
          for k in range(n):
            A[j][k]-=multiple*A[i][k]
            I[j][k]-=multiple*I[i][k]
    return I
```

```python
import time
import random
import numpy as np
import matplotlib.pyplot as plt

# Your functions assumed defined:
# invSolve(A)
# invSolve2(A)

def random_matrix(n):
    return [[random.random() for _ in range(n)] for _ in range(n)]

sizes = [2, 4, 8, 12, 16, 20, 24, 28, 32]

times_custom = []
times_numpy = []

for n in sizes:
    runs = 5
    total_custom = 0
    total_numpy = 0

    for _ in range(runs):
        A = random_matrix(n)

        # ---- Custom inverse ----
        start = time.perf_counter()
        invSolve(A)
        end = time.perf_counter()
        total_custom += (end - start)

        # ---- NumPy inverse ----
        A_np = np.array(A)
        start = time.perf_counter()
        invSolve2(A_np)
        end = time.perf_counter()
        total_numpy += (end - start)

    times_custom.append(total_custom / runs)
    times_numpy.append(total_numpy / runs)

    print(f"n={n} | custom={times_custom[-1]:.6f} | numpy={times_numpy[-1]:.6f}")

# ---- Plot both on same graph ----
plt.figure()
plt.plot(sizes, times_custom)
plt.plot(sizes, times_numpy)
plt.xlabel("Matrix size (n)")
plt.ylabel("Time (seconds)")
plt.title("Custom Gauss-Jordan vs NumPy Inverse")
plt.legend(["Custom Gauss-Jordan", "NumPy"])
plt.show()
```

## The algorithm based on the paper (we use the AX=B soln given by the paper and then find inverses by passing I instead of b)

```python
def quantum_inspired_solve(A, b, iterations=1000):
    """
    Solve Ax = b using randomized Kaczmarz (quantum-inspired).
    """

    m, n = A.shape
    x = np.zeros(n)

    # Precompute row norms (important for sampling!)
    row_norms = np.linalg.norm(A, axis=1)**2
    probs = row_norms / np.sum(row_norms)

    for _ in range(iterations):
        i = np.random.choice(m, p=probs)
        Ai = A[i]
        
        # Kaczmarz update
        x = x + (b[i] - Ai @ x) * Ai / row_norms[i]

    return x
```

```python
def inverse_via_solver(A):
    n = A.shape[0]
    I = np.eye(n)
    A_inv = np.zeros_like(A)

    for i in range(n):
        A_inv[:, i] = quantum_inspired_solve(A, I[:, i])

    return A_inv
```

```python
def quantum_inspired_inverse(A, rank_k=10):
    """
    Approximate inverse using randomized low-rank projection.
    Works best when A has low-rank structure.
    """
    A = np.array(A)
    n = A.shape[0]

    # Random projection
    Omega = np.random.randn(n, rank_k)
    Y = A @ Omega

    # Orthonormal basis
    Q, _ = np.linalg.qr(Y)

    # Reduced matrix
    B = Q.T @ A @ Q

    # Invert reduced system
    B_inv = np.linalg.inv(B)

    # Lift back
    A_inv_approx = Q @ B_inv @ Q.T

    return A_inv_approx
```

```python
import time
import random
import numpy as np
import matplotlib.pyplot as plt

# Your functions assumed defined:
# invSolve(A)
# invSolve2(A)

def random_matrix(n):
    return [[random.random() for _ in range(n)] for _ in range(n)]

sizes = [2, 4, 8, 12, 16, 20, 24, 28, 32, 40,64,80,100]

times_custom = []
times_numpy = []
times_quantum=[]
times_quantum_2=[]
for n in sizes:
    runs = 5
    total_custom = 0
    total_numpy = 0
    total_quantum = 0
    total_quantum_2=0

    for _ in range(runs):
        A = random_matrix(n)

        # ---- Custom inverse ----
        start = time.perf_counter()
        invSolve(A)
        end = time.perf_counter()
        total_custom += (end - start)

        # ---- NumPy inverse ----
        A_np = np.array(A)
        start = time.perf_counter()
        invSolve2(A_np)
        end = time.perf_counter()
        total_numpy += (end - start)

        # ---- Quantum-inspired inverse ----
        start = time.perf_counter()
        quantum_inspired_inverse(A_np, rank_k=min(10, n))
        end = time.perf_counter()
        total_quantum += (end - start)

        # -----Quantum inspired inverse from direct implementation of paper-----
        start=time.perf_counter()
        inverse_via_solver(A_np)
        end=time.perf_counter()
        total_quantum_2+=(end-start)
        
    times_custom.append(total_custom / runs)
    times_numpy.append(total_numpy / runs)
    times_quantum.append(total_quantum / runs)
    times_quantum_2.append(total_quantum_2/runs)

    print(f"n={n} | custom={times_custom[-1]:.6f} | numpy={times_numpy[-1]:.6f} | quantum={times_quantum[-1]:.6f}")

plt.figure()
plt.plot(sizes, times_custom)
plt.plot(sizes, times_numpy)
plt.plot(sizes, times_quantum)

plt.xlabel("Matrix size (n)")
plt.ylabel("Time (seconds)")
plt.title("Inverse Comparison: Custom vs NumPy vs Quantum-Inspired")

plt.legend([
    "Custom Gauss-Jordan",
    "NumPy (linalg.inv)",
    "Quantum-Inspired Approx",
])

plt.show()
```
